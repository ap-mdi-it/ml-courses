# Datatypes en containers

In deze notebook verkennen we de basis van Python: de ingebouwde datatypes en de vier container-types die je voortdurend zult gebruiken in ML.

## Basis datatypes

### Getallen

Python heeft twee numerieke types die je constant tegenkomt: `int` (geheel getal) en `float` (kommagetal). Het type van een waarde kun je opvragen met `type()`.

```python
x = 10
y = 3.14

print(type(x))  # <class 'int'>
print(type(y))  # <class 'float'>
```

```python
# Arithmetic operators
print(x + 2)  # addition
print(x - 3)  # subtraction
print(x * 4)  # multiplication
print(x / 3)  # division (always returns float)
print(x // 3)  # integer division
print(x % 3)  # modulo (remainder)
print(x**2)  # exponentiation
```

```python
# Compound assignment
x += 1
print(x)  # 11
x *= 2
print(x)  # 22
```

### Booleans

**Booleans** (`bool`) kunnen slechts twee waarden aannemen: `True` of `False`. Ze zijn het resultaat van vergelijkingen en worden gebruikt in condities.

```python
a = True
b = False

print(a and b)  # False — both must be True
print(a or b)  # True  — at least one must be True
print(not a)  # False
print(a != b)  # True  — exclusive OR (XOR)
```

```python
# Comparison operators return booleans
print(5 > 3)  # True
print(5 == 5)  # True
print(5 != 4)  # True
print(5 >= 10)  # False
```

### Strings

**Strings** zijn reeksen van tekens. Je kunt ze definiëren met enkele of dubbele aanhalingstekens. **F-strings** (formatted string literals) zijn de meest leesbare manier om waarden in een string in te voegen.

```python
name = "Alice"
age = 22

# F-string: insert values directly into text
print(f"{name} is {age} years old.")
```

```python
# Useful string methods
s = "  Hello, World!  "

print(s.strip())  # remove leading/trailing whitespace
print(s.strip().lower())  # convert to lowercase
print(s.strip().upper())  # convert to uppercase
print(s.strip().replace("World", "Python"))  # replace substring
print(s.strip().split(","))  # split on character
```

```python
# Strings can be indexed and sliced (like lists)
word = "python"
print(word[0])  # first character: 'p'
print(word[-1])  # last character: 'n'
print(word[1:4])  # slice: 'yth'
print(len(word))  # length: 6
```

## Containers

Python heeft vier ingebouwde container-types om meerdere waarden bij te houden:

| Type | Geordend | Muteerbaar | Uniek | Gebruik |
|------|---------|-----------|-------|--------|
| `list` | ✓ | ✓ | ✗ | Reeksen van elementen |
| `dict` | ✓ | ✓ | *keys* | *key-value* paren |
| `set` | ✗ | ✓ | ✓ | Unieke elementen, lidmaatschapstest |
| `tuple` | ✓ | ✗ | ✗ | Onveranderlijke reeksen |

### List

Een **list** is een geordende, muteerbare reeks van elementen. Elementen mogen van verschillende types zijn.

```python
scores = [85, 92, 78, 95, 88]

print(scores[0])  # first element: 85
print(scores[-1])  # last element: 88
print(scores[1:3])  # slice: [92, 78]
print(len(scores))  # length: 5
```

```python
# Adding and removing elements
scores.append(100)  # add to the end
removed = scores.pop()  # remove and return the last element
print(removed)  # 100
print(scores)
```

```python
# Iterating over a list
animals = ["cat", "dog", "monkey"]
for animal in animals:
    print(animal)

# With index via enumerate()
for i, animal in enumerate(animals):
    print(f"#{i + 1}: {animal}")
```

```python
# List comprehension: elegant one-liner for building lists
squares = [x**2 for x in range(6)]
print(squares)  # [0, 1, 4, 9, 16, 25]

# With a condition
even_squares = [x**2 for x in range(10) if x % 2 == 0]
print(even_squares)  # [0, 4, 16, 36, 64]
```

### Dictionary

Een **dictionary** (`dict`) slaat *key-value* paren op. Je kunt een waarde opvragen via de bijhorende *key*. *Keys* moeten uniek en onveranderlijk zijn.

```python
student = {
    "name": "Alice",
    "age": 22,
    "score": 91.5,
}

print(student["name"])  # 'Alice'
print("score" in student)  # True — membership test
print(student.get("year", "?"))  # '?' — safe lookup with default value
```

```python
# Add, modify and delete
student["major"] = "AI"  # add new key-value pair
student["score"] = 93.0  # modify value
del student["age"]  # delete

print(student)
```

```python
# Iterate over key-value pairs
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Dictionary comprehension
words = ["apple", "banana", "cherry"]
word_lengths = {w: len(w) for w in words}
print(word_lengths)  # {'apple': 5, 'banana': 6, 'cherry': 6}
```

### Set

Een **set** is een ongeordende verzameling van *unieke* elementen. Ideaal voor het snel controleren van lidmaatschap en het verwijderen van duplicaten.

```python
colors = {"red", "green", "blue"}

print("red" in colors)  # True
print("yellow" in colors)  # False

colors.add("yellow")
colors.add("red")  # duplicate is ignored
print(colors)
```

```python
# Remove duplicates from a list via set()
values = [1, 2, 2, 3, 3, 3, 4]
unique = list(set(values))
print(unique)
```

### Tuple

Een **tuple** is een geordende, *onveranderlijke* reeks. Eens aangemaakt kunnen elementen niet meer gewijzigd worden. Tuples worden veel gebruikt om meerdere waarden tegelijk terug te geven vanuit een functie.

```python
point = (3.5, -2.0)  # x, y coordinates

x, y = point  # tuple unpacking
print(f"x={x}, y={y}")

# Tuples can be used as dictionary keys (lists cannot!)
locations = {(0, 0): "origin", (1, 0): "right"}
print(locations[(0, 0)])
```

### Type-conversie

Je kunt expliciet omzetten tussen types met `int()`, `float()`, `str()`, `list()`, `set()` en `tuple()`.

```python
print(int(3.9))  # 3 (truncates, does not round)
print(float("3.14"))  # 3.14
print(str(42))  # '42'
print(list({1, 2, 3}))  # [1, 2, 3] — order not guaranteed
print(tuple(["a", "b", "c"]))  # ('a', 'b', 'c')

# isinstance() checks the type
print(isinstance(3.14, float))  # True
print(isinstance("hello", (str, int)))  # True — checks against multiple types
```
