# Controle-structuren en comprehensions

Programma's voeren stappen uit in een bepaalde volgorde. Met **controle-structuren** stuur je die volgorde: je voert code conditioneel uit, herhaalt ze, of slaat ze over. In deze notebook leer je ook **comprehensions** en **lambda-functies** — twee compacte schrijfwijzen die je voortdurend tegenkomt in ML code.

## Condities: `if / elif / else`

Met een **if-statement** voer je een blok code alleen uit als een conditie `True` is.

````python
score = 74

if score >= 90:
    print("Distinction")
elif score >= 60:
    print("Passed")
else:
    print("Failed")
````

````python
# Ternary expression — compact if/else on one line
status = "passed" if score >= 60 else "failed"
print(status)
````

## Lussen

### `for`-lus

Een **for-lus** itereert over een reeks: een lijst, string, dictionary, `range()`, enz.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

````python
# range(start, stop, step)
for i in range(0, 10, 2):  # 0, 2, 4, 6, 8
    print(i, end=" ")
print()  # newline
````

````python
# enumerate() gives index + value simultaneously
for i, fruit in enumerate(fruits, start=1):
    print(f"{i}. {fruit}")
````

````python
# zip() pairs two sequences element by element
names = ["Alice", "Bob", "Charlie"]
scores = [88, 74, 95]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

### `while`-lus

Een **while-lus** blijft herhalen zolang een conditie `True` is. Gebruik dit wanneer je het aantal iteraties op voorhand niet kent.

````python
counter = 1
while counter <= 5:
    print(counter)
    counter += 1
````

### `break` en `continue`

- **`break`**: beëindigt de lus onmiddellijk
- **`continue`**: slaat de rest van de huidige iteratie over en gaat naar de volgende

````python
# Find the first even number greater than 10
for n in range(11, 30):
    if n % 2 == 0:
        print(f"First even number > 10: {n}")
        break
````

````python
# Skip odd numbers
for n in range(1, 11):
    if n % 2 != 0:
        continue
    print(n, end=" ")
print()
````

## Comprehensions

**Comprehensions** zijn een compacte en leesbare manier om lijsten, dictionaries of sets te bouwen op basis van een reeks. Ze vervangen een `for`-lus + `append()`-patroon door één expressie.

```
[expression for element in sequence if condition]
```

````python
# List comprehension
squares = [x**2 for x in range(1, 8)]
print(squares)
````

````python
# Filter and transform in one step
words = ["data", "science", "AI", "ML", "python", "notebook"]
long_words = [w.upper() for w in words if len(w) > 4]
print(long_words)
````

````python
# Dictionary comprehension
names = ["Alice", "Bob", "Charlie"]
scores = [88, 74, 95]
report = {name: score for name, score in zip(names, scores)}
print(report)
````

````python
# Set comprehension — automatically unique values
text = "programming is fun and programming is useful"
unique_words = {w for w in text.split()}
print(unique_words)
````

## Lambda-functies

Een **lambda** is een anonieme functie die in één expressie geschreven wordt. Gebruik ze voor eenvoudige, éénmalige transformaties — typisch als argument voor `sorted()`, `map()` of `.apply()` in Pandas.

```python
lambda arguments: expression
```

````python
# Simple lambda
double = lambda x: x * 2
print(double(5))  # 10
````

````python
# Lambda as sort key
students = [("Alice", 88), ("Bob", 74), ("Charlie", 95), ("Diana", 81)]

# Sort by score (second element of each tuple)
ranked = sorted(students, key=lambda s: s[1], reverse=True)
for name, score in ranked:
    print(f"{name}: {score}")
````

````python
# Lambda with map() — apply a function to each element
scores = [85, 92, 71, 68, 95]
normalized = list(map(lambda s: round(s / 100, 2), scores))
print(normalized)
````
