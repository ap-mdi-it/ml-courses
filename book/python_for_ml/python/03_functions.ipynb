# Functies en modules

**Functies** zijn herbruikbare blokken code. Ze nemen invoer aan, voeren een berekening uit en geven een resultaat terug. In deze notebook leer je functies definiëren, ze flexibel maken met parameters, en bestaande bibliotheken importeren.

## Functies definiëren

Een functie definieer je met het sleutelwoord `def`. De **docstring** (eerste string in de functie) beschrijft wat de functie doet.

```python
def greet(name):
    """Return a greeting for the given name."""
    return f"Hello, {name}!"


print(greet("Alice"))
print(greet("Bob"))
```

## Parameters en standaardwaarden

Je kunt parameters een **standaardwaarde** (of *default*-waarde) geven. Die waarde wordt gebruikt wanneer de aanroeper het argument weglaat.

```python
def calculate_bmi(weight_kg, height_m, verbose=True):
    """Calculate the Body Mass Index.

    Parameters
    ----------
    weight_kg : float
        Weight in kilograms.
    height_m : float
        Height in metres.
    verbose : bool, optional
        If True, also print the category. Default True.

    Returns
    -------
    float
        BMI rounded to one decimal place.
    """
    bmi = weight_kg / height_m**2
    if verbose:
        if bmi < 18.5:
            category = "underweight"
        elif bmi < 25:
            category = "normal weight"
        elif bmi < 30:
            category = "overweight"
        else:
            category = "obese"
        print(f"BMI: {bmi:.1f} ({category})")
    return round(bmi, 1)


calculate_bmi(70, 1.75)
bmi_value = calculate_bmi(90, 1.80, verbose=False)
print(bmi_value)
```

## Meerdere returnwaarden

Python-functies kunnen meerdere waarden tegelijkertijd teruggeven via een **tuple**.

```python
def summarize(values):
    """Return the minimum, maximum and mean of a list of numbers."""
    return min(values), max(values), sum(values) / len(values)


data = [4, 7, 2, 9, 1, 8, 5]
minimum, maximum, mean = summarize(data)  # tuple unpacking
print(f"min={minimum}, max={maximum}, mean={mean:.2f}")
```

## `*args` en `**kwargs`

- **`*args`** laat een functie een willekeurig aantal positionele argumenten aanvaarden (als tuple).
- **`**kwargs`** laat een functie een willekeurig aantal sleutelwoordargumenten aanvaarden (als dictionary).

Je ziet dit patroon vaak in bibliotheken als Pandas en Matplotlib.

```python
def total(*numbers):
    """Sum all provided numbers."""
    return sum(numbers)


print(total(1, 2, 3))  # 6
print(total(10, 20, 30, 40))  # 100
```

```python
def print_info(**attributes):
    """Print keyword arguments as key-value properties."""
    for key, value in attributes.items():
        print(f"  {key}: {value}")


print_info(name="Alice", age=22, major="AI")
```

## Scope

Variabelen die binnenin een functie worden aangemaakt, zijn **lokaal** — ze bestaan niet buiten de functie. Variabelen buiten functies zijn **globaal**.

```python
threshold = 60  # global variable


def is_passing(score):
    result = score >= threshold  # local variable + reference to global
    return result


print(is_passing(75))  # True
print(is_passing(45))  # False
# print(result)  # NameError: 'result' does not exist outside the function
```

## Hulpfuncties schrijven voor data-analyse

In de praktijk schrijf je regelmatig kleine hulpfuncties om herhaalde stappen te bundelen. Hier zijn twee realistische voorbeelden:

```python
def normalize(values):
    """Scale values to the interval [0, 1] using min-max normalization.

    Parameters
    ----------
    values : list of float
        The values to normalize.

    Returns
    -------
    list of float
        Normalized values between 0 and 1.
    """
    min_val = min(values)
    max_val = max(values)
    return [(v - min_val) / (max_val - min_val) for v in values]


data = [10, 20, 30, 40, 50]
print(normalize(data))  # [0.0, 0.25, 0.5, 0.75, 1.0]
```

```python
def clean_column_name(text):
    """Strip whitespace, lowercase and replace spaces with underscores.

    Parameters
    ----------
    text : str

    Returns
    -------
    str
        Cleaned string suitable as a column name.
    """
    return text.strip().lower().replace(" ", "_")


column_names = ["Name Student", "  Age  ", "Score (%)"]
clean_names = [clean_column_name(c) for c in column_names]
print(clean_names)
```
