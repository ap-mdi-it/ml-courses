# Pandas — Selectie en filtering

In deze notebook leer je de meest gebruikte technieken om rijen en kolommen te selecteren uit een DataFrame:

- Kolomselectie
- **Boolean filtering** — de primaire manier om rijen te filteren
- **`.loc[]`** en **`.iloc[]`** — voor label- en positie-gebaseerde selectie
- Sorteren en tellen

Alle technieken worden intensief gebruikt in de latere ML-notebooks.

```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
titanic.head()
```

## Kolomselectie

```python
# One column → Series
ages = titanic["age"]
print(type(ages))
ages.head()
```

```python
# Multiple columns → DataFrame (double square brackets)
selected = titanic[["survived", "pclass", "age", "fare"]]
selected.head()
```

## Boolean filtering

Filtering werkt net als bij NumPy: je maakt een masker van `True`/`False`-waarden en past het toe op het DataFrame.

Combineer condities met:
- `&` (AND)
- `|` (OR)
- `~` (NOT)

⚠️ Zet elke conditie tussen haakjes wanneer je combineert.

```python
# Simple condition
women = titanic[titanic["sex"] == "female"]
print(f"Number of women: {len(women)}")
women.head()
```

```python
# Combine multiple conditions with &
first_class_survivors = titanic[(titanic["survived"] == 1) & (titanic["pclass"] == 1)]
print(f"First-class survivors: {len(first_class_survivors)}")
```

```python
# Combine with | — passengers in 1st or 2nd class
first_or_second = titanic[(titanic["pclass"] == 1) | (titanic["pclass"] == 2)]
print(len(first_or_second))

# Alternative with .isin()
first_or_second_v2 = titanic[titanic["pclass"].isin([1, 2])]
print(len(first_or_second_v2))
```

```python
# Comparison with numeric threshold
minors = titanic[titanic["age"] < 18]
print(f"Minors on board: {len(minors)}")
```

## `.loc[]` — label-gebaseerde selectie

Met **`.loc[rows, columns]`** selecteer je op basis van **indexlabels** en **kolomnamen**.

```python
# Select rows 5 through 9, columns 'survived', 'age' and 'sex'
titanic.loc[5:9, ["survived", "age", "sex"]]
```

```python
# .loc with boolean mask + column selection
# Returns surviving women with their age and class
titanic.loc[
    (titanic["survived"] == 1) & (titanic["sex"] == "female"), ["age", "pclass", "fare"]
].head()
```

## `.iloc[]` — positie-gebaseerde selectie

Met **`.iloc[rows, columns]`** selecteer je op basis van **gehele getallen** (positie, 0-gebaseerd). Handig wanneer je de kolomnamen niet kent.

```python
# First 5 rows, columns 0 through 3
titanic.iloc[0:5, 0:4]
```

```python
# Last row
titanic.iloc[-1]
```

## Sorteren

```python
# Sort by one column (descending)
titanic.sort_values("fare", ascending=False).head()
```

```python
# Sort by multiple columns
titanic.sort_values(["pclass", "fare"], ascending=[True, False]).head(8)
```

## Unieke waarden tellen

```python
# Unique values in a column
print(titanic["pclass"].unique())
print(titanic["sex"].nunique(), "unique values in 'sex'")
```

```python
# Count occurrences
print(titanic["pclass"].value_counts())
print()
# As proportions
print(titanic["survived"].value_counts(normalize=True).round(2))
```

## Index resetten

Na filtering klopt de index van het resulterende DataFrame niet meer aaneengesloten. Gebruik `.reset_index(drop=True)` om de index opnieuw te nummeren.

```python
women = titanic[titanic["sex"] == "female"].reset_index(drop=True)
print(women.index[:5].tolist())  # [0, 1, 2, 3, 4]
```
