# Pandas — DataFrames aanmaken en inspecteren

**Pandas** is de standaardbibliotheek voor het werken met tabelgegevens in Python. De twee centrale datastructuren zijn:

- **`DataFrame`**: een tweedimensionale tabel met benoemde kolommen en een index (vergelijkbaar met een spreadsheet of SQL-tabel)
- **`Series`**: een eendimensionale gelabelde array — elke kolom van een DataFrame is een Series

In deze notebook leer je DataFrames aanmaken, inladen en inspecteren.

```python
import numpy as np
import pandas as pd

print(pd.__version__)
```

## DataFrame aanmaken

### Vanuit een dictionary

```python
data = {
    "name": ["Alice", "Bob", "Charlie", "Diana"],
    "age": [22, 25, 23, 21],
    "score": [88.5, 74.0, 92.3, 81.7],
    "passed": [True, True, True, True],
}

df = pd.DataFrame(data)
df
```

### Een CSV-bestand inladen

In de praktijk lees je data bijna altijd uit een bestand of een externe bron. Hier gebruiken we de **Titanic-dataset** van Seaborn — een klassieker in data-analyse.

```python
import seaborn as sns

# Load the Titanic dataset (automatically downloaded on first run)
titanic = sns.load_dataset("titanic")

# Or from a CSV file:
# df = pd.read_csv("path/to/file.csv")

titanic.head()
```

## Een DataFrame inspecteren

Dit zijn de eerste stappen die je bij **elke** nieuwe dataset uitvoert.

```python
# View first and last rows
print(titanic.head(5))  # first 5 rows
print()
print(titanic.tail(3))  # last 3 rows
```

```python
# Dimensions
print("shape:", titanic.shape)  # (rows, columns)
print("rows: ", len(titanic))
```

```python
# Column names
print(titanic.columns.tolist())
```

```python
# Data type of each column
print(titanic.dtypes)
```

```python
# Concise summary: types + non-null counts
titanic.info()
```

```python
# Statistical summary of numeric columns
titanic.describe()
```

```python
# Include categorical columns — .T transposes for readability
titanic.describe(include="all").T
```

## Series vs. DataFrame

Een **Series** is een eendimensionale kolom; een **DataFrame** is een verzameling Series die een gemeenschappelijke index delen.

```python
# Selecting one column returns a Series
ages = titanic["age"]
print(type(ages))  # <class 'pandas.core.series.Series'>
print(ages.dtype)  # float64
print(ages.head())
```

```python
# Selecting multiple columns returns a DataFrame
selected = titanic[["survived", "age", "fare"]]
print(type(selected))  # <class 'pandas.core.frame.DataFrame'>
selected.head()
```

```python
# Access the underlying NumPy array
print(ages.values[:5])  # numpy.ndarray
print(type(ages.values))
```

## Ontbrekende waarden snel verkennen

```python
# Number of missing values per column
print(titanic.isnull().sum())
```

```python
# Percentage of missing values
pct = titanic.isnull().mean() * 100
print(pct[pct > 0].round(1))  # show only columns with missing values
```
